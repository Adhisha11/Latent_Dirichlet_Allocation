### 1. Document Preprocessing and Representation

We will define a sample corpus of documents, preprocess the text (lowercasing, tokenization, removing stop words), and convert the documents into a document-term matrix using bag-of-words representation.

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Sample corpus representing different themes (Technology, Food, Space)
corpus = [
    "Artificial intelligence and machine learning are shaping the future of technology.",
    "Deep neural networks and computer vision are key components of modern AI.",
    "The chef prepared a delicious three-course meal featuring fresh pasta and organic ingredients.",
    "I love baking homemade sourdough bread and pastry desserts in my kitchen.",
    "NASA launched a new space telescope to study distant galaxies and black holes.",
    "The exploration of Mars and lunar landing missions are crucial for space science.",
    "Software developers write code in Python and JavaScript to build web applications.",
    "A balanced diet rich in vegetables, fruits, and proteins is essential for health."
]

# Preprocess and create a Document-Term Matrix (Bag of Words)
# We'll use scikit-learn's CountVectorizer to tokenize and remove standard English stop words.
vectorizer = CountVectorizer(stop_words='english')
doc_term_matrix = vectorizer.fit_transform(corpus)

# Convert to a DataFrame for visualization
feature_names = vectorizer.get_feature_names_out()
df_dtm = pd.DataFrame(doc_term_matrix.toarray(), columns=feature_names)
display(df_dtm.head())
```

### 2. LDA Model Training

We will train the Latent Dirichlet Allocation model on our document-term matrix to discover latent topics.

```python
# We fit LDA with 3 topics representing Technology, Food, and Space
n_topics = 3
lda_model = LatentDirichletAllocation(n_components=n_topics, random_state=42)
lda_model.fit(doc_term_matrix)

print("LDA model training completed successfully.")
```

### 3. Topic-Word Distributions

Let's extract and display the top words for each discovered topic to interpret what each topic represents.

```python
def display_topics(model, feature_names, no_top_words):
    for topic_idx, topic in enumerate(model.components_):
        print(f"Topic {topic_idx + 1}:")
        # Get top words based on their weight in the topic
        top_features_ind = topic.argsort()[:-no_top_words - 1:-1]
        top_features = [feature_names[i] for i in top_features_ind]
        weights = [topic[i] for i in top_features_ind]

        # Display word-weight pairs
        for word, weight in zip(top_features, weights):
            print(f"  - {word}: {weight:.2f}")
        print()

no_top_words = 5
display_topics(lda_model, feature_names, no_top_words)
```
